# 02 ? Convolutional Neural Networks (CNN) Foundations

In Chapter 01 (`Basis.ipynb`), we built an MLP with `nn.Linear` layers. For image data, flattening loses spatial relationships and explodes the parameter count.

**Convolutional Neural Networks (CNNs)** solve this by using small sliding filters that preserve 2D grid structure, share weights across spatial positions, and detect local patterns (edges, corners, textures).

## 1. Convolution Arithmetic: Kernel, Stride & Padding

A 2D convolution applies a learnable filter (kernel) across spatial positions:
- **`in_channels` ($C_{in}$)**: Number of input channels (e.g. 1 for grayscale, 3 for RGB).
- **`out_channels` ($C_{out}$)**: Number of distinct feature detectors to learn.
- **`kernel_size` ($K$)**: Dimensions of the sliding window (typically $3\times 3$ or $5\times 5$).
- **`stride` ($S$)**: Step size the kernel moves by (default $1$).
- **`padding` ($P$)**: Extra zeros added around input borders to preserve spatial size.

---

### The Output Size Formula

Given an input of spatial size $W_{in} \times H_{in}$:
$$\text{dim}_{out} = \left\lfloor \frac{\text{dim}_{in} - K + 2P}{S} \right\rfloor + 1$$


```python
import torch
import torch.nn as nn

# 1. Simulate a batch of 4 grayscale images (FashionMNIST shape: 28x28)
# Shape: (Batch, Channels, Height, Width) -> (B, C, H, W)
x = torch.rand(4, 1, 28, 28)
print("Input batch shape:", x.shape)

# 2. Define a Conv2d layer:
# - in_channels=1 (grayscale)
# - out_channels=16 (extract 16 distinct feature maps)
# - kernel_size=3 (3x3 filter)
# - stride=1
# - padding=1 (keeps spatial dimensions 28x28 when kernel=3, stride=1)
conv1 = nn.Conv2d(in_channels=1, out_channels=16, kernel_size=3, stride=1, padding=1)

out = conv1(x)
print("Output feature map shape:", out.shape)  # Expected: (4, 16, 28, 28)

# 3. Inspect learned parameters (weights & bias)
# Kernel weight shape: (out_channels, in_channels, kernel_h, kernel_w)
print("Filter weights shape:", conv1.weight.shape)  # (16, 1, 3, 3)
print("Bias shape:", conv1.bias.shape)              # (16,)
print(f"Total parameters in layer: {conv1.weight.numel() + conv1.bias.numel()}")
```
